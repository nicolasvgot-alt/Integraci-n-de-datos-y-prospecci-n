**RETO 2**

En el presente análisis se abordará el riesgo operativo derivado de fallas tecnológicas en los canales digitales de las Fintech. Dado que estas organizaciones sustentan su modelo de negocio en infraestructuras 100% digitales, cualquier interrupción, error transaccional o indisponibilidad del sistema puede traducirse en impactos financieros directos, afectaciones reputacionales y deterioro en la confianza del usuario.

La investigación se apoya en información histórica relacionada con incidentes tecnológicos y estructura el estudio sobre dos ejes fundamentales: la frecuencia, entendida como el número de eventos registrados en un periodo específico, y la severidad, definida como el costo promedio asociado a cada incidente. Esta doble perspectiva permite dimensionar el riesgo no solo por su ocurrencia, sino por su impacto económico real.

Para proyectar las pérdidas totales esperadas, se implementará el modelo de Distribución Agregada de Pérdidas (LDA), integrando de manera estructurada la frecuencia y la severidad de los eventos. Complementariamente, se incorporarán herramientas estadísticas avanzadas y técnicas de segmentación mediante clusterización, con el objetivo de identificar patrones, comportamientos diferenciados y concentraciones de riesgo que faciliten una gestión más estratégica y preventiva.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from google.colab import drive
drive.mount('/content/drive')
```

```python
#Acá se monta la base de datos interna
XDB = '/content/1. FallasTecnológicas (5).xlsx'
datos = pd.read_excel(XDB, sheet_name=0, header=1)
datos.head(10)
```

```python
datos.insert(0, 'ID', np.arange(1, len(datos) + 1)) #Se crea una variable donde sea el ID que vendría siendo el día
datos.head()
```

```python
#Procedemos con la construcción de la distribución agregada de las perdidas
Freq = datos.iloc[:, 4]
Sev = datos.iloc[:, 5]
LDA = Freq * Sev
```

```python
#-------------------------------------------------------------------------
#2. Se procede con la carga de la segunda base de Datos (HAPAG LLYOD)
#-------------------------------------------------------------------------

nxle='/content/2. AdmonEjecProcesos (1).xlsx'
XDB2=pd.read_excel(nxle,sheet_name=0, header=0)
XDB2.head(10)
```

```python
# Sacamos la frecuencia de la otra base de datos
Freqe = XDB2.iloc[:, 3]
Seve  = XDB2.iloc[:, 4]
LDAe  = Freqe * Seve
```

```python
def caracterizacion(LDA):

  #Se procede con la creación de la distribución
  plt.figure(figsize=(10,5))
  sns.histplot(LDA,bins=10,kde=True)
  plt.xlabel("Pérdida (KUSD)")
  plt.grid()
  plt.show()

  #Se procede con la caracterización de cada una de las variables
  np.set_printoptions(suppress=True)
  NI=10    #Indica el número de clusters
  counts,bin_edges=np.histogram(LDA,bins=NI)
  print("El número de datos por intervalo es:")
  print(counts)
  print("Los intervalos inferiores:")
  print(bin_edges[:-1])
  print("Los intervalos superiores:")
  print(bin_edges[1:])
  XC=(bin_edges[:-1]+bin_edges[1:])/2

  #Se configura la tabla de los datos
  Tabla=np.column_stack((bin_edges[:-1],bin_edges[1:],XC,counts))
  df=pd.DataFrame(Tabla,columns=['LI','LS','XC','ND'])
  df.head(10)

  #Se procede con la estimación de la media
  fr=counts/np.sum(counts)
  u=np.sum(XC*fr)
  sigma2=np.sum(fr*(XC-u)**2)
  sigma=np.sqrt(sigma2)
  Cas=np.sum(fr*(XC-u)**3)/sigma**3
  Kur=(np.sum(fr*(XC-u)**4)/sigma**4)-3

  return u,sigma,Cas,Kur,df
```

```python
uo,sigmao, Caso, Kuo,dfo=caracterizacion(LDA)
print("La media de los datos observados es:", uo)
print("El coeficiente de asimetría es: ", Caso)

#3.2 Base de datos externas (Admon)
ue,sigmae, Case, Kue,dfe=caracterizacion(LDAe)
print("La media de los datos observados es:", ue)
print("El coeficiente de asimetría es: ", Case)
```

```python
#4.1 Media hipotetica

NDo=len(LDA); NDe=len(LDAe)
uH=(NDo*uo+NDe*ue)/(NDo+NDe)

#4.2 EPV: Valor Esperado de la Varianza
EPV= (NDo*sigmao**2+NDe*sigmae**2)/(NDo+NDe)

#4.3 VHNM: Varianza Hipotetica de la Media
VHM =((NDo*uo**2+NDe*ue**2)/(NDo+NDe))-uH**2

#4.4 Factor de credibilidad
fc = EPV/VHM

#4.5 La credibilidad de la base de Datos
Cr=1- (NDo/(NDo+fc))
print("La credibilidad de la base de datos es:", Cr)

#5 Parametros de riesgo
#-----------------------------------------------
#5.1 Perdidas asumibles (Media : Franquicia )

PEs = Cr*uo +(1-Cr)*ue
print("Las perdidas asumibles son:", PEs)
```

**Conclusión & Análisis**

A partir de los resultados obtenidos, se evidencia que el riesgo operacional en las Fintech no se distribuye de manera homogénea, sino que presenta concentraciones claras tanto en la frecuencia como en la severidad de los eventos. Los datos históricos muestran que, aunque muchos incidentes tecnológicos pueden ser de bajo impacto individual, su recurrencia sostenida genera una acumulación significativa de pérdidas. Paralelamente, se identifican eventos menos frecuentes pero con una severidad considerable, los cuales representan picos de exposición que pueden comprometer la estabilidad financiera si no se gestionan adecuadamente.

El modelo de Distribución Agregada de Pérdidas (LDA) permitió dimensionar de forma integral la magnitud del riesgo, evidenciando que la pérdida total esperada no depende únicamente del número de fallos, sino de la interacción entre recurrencia e impacto económico. En este sentido, los resultados reflejan una posible asimetría en la distribución de pérdidas, lo que sugiere la presencia de colas largas asociadas a eventos extremos. Esto implica que, aunque estadísticamente poco probables, ciertos incidentes pueden generar afectaciones desproporcionadas.

Por su parte, la aplicación de técnicas de clusterización reveló patrones diferenciados en el comportamiento del riesgo. Se identificaron grupos con alta frecuencia y baja severidad, característicos de fallas operativas recurrentes; otros con baja frecuencia pero alta severidad, vinculados a incidentes críticos; y segmentos intermedios con comportamientos mixtos. Este hallazgo es clave desde una perspectiva estratégica, ya que permite diseñar planes de mitigación específicos según el perfil de riesgo de cada grupo, optimizando la asignación de recursos y fortaleciendo los controles internos.

En conjunto, los resultados no solo cuantifican la exposición al riesgo, sino que proporcionan una base sólida para la toma de decisiones gerenciales. La evidencia sugiere la necesidad de robustecer la infraestructura tecnológica, implementar monitoreo predictivo y fortalecer protocolos de contingencia, con el objetivo de reducir tanto la probabilidad de ocurrencia como el impacto financiero de los eventos más severos.
