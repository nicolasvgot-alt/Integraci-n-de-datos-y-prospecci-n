# Riesgo operacional - EVERGREEN
En este espacio se plantea como caso de estuido los riesogos asociaosos a las operaciones del negocio de la Naviera EVERGREEN. Esta Naviera Taiwanesa opera en 240 puertos alrededor del mundo utilizando 150 buques canal de Panamá
Los riesgos a los cuales esta expuesta la naviera estan relacionaods con as logistica portuaria frente a la carga y descarga de contenedores. De esta manera, este riesgo operacional se describe a través de 2 variables:

Frecuencia: Indica el numero de contenedores que se quedan en tierra x semana(Tiempo)
Severidad: Indica el costo de solucionar cada evento de riesgo
Es importante mencionar que el producto entre la severidad y la frecuencia da como resultado la distribución agregada de las pérdidas.
Es precisamanete que se hará la caracterización utilizando las medidas de tendencia central (Media, Desviación) y las medidas de dispersión (Coeficiente de asimetria, Kurtosis)

```python
# Procedemos con la construcción de la distribución agregada de las pérdidas
Freq=XDB.iloc[:,3]
sev=XDB.iloc[:,4]
LDA=Freq*sev

#Se procede con la construcción de la distribución agregada de las pérdidas
plt.figure(figsize=(10,5))
sns.histplot(LDA,bins=10,kde=True)
plt.xlabel("Pérdidas USA")
plt.grid()
plt.show()
```
